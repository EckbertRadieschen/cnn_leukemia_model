```python
import sys

import tensorflow as tf
import keras
import numpy as np
import matplotlib.pyplot as plt
import time
from zipfile import ZipFile

from pathlib import Path

BASE_DIR = Path().cwd().parent

sys.path.append(str(BASE_DIR))
from functions import extract_picture_file, vectorize_image, predict_picture

PICTURE_DATA_DIR = BASE_DIR / "picture_data"

picture_zip_path = PICTURE_DATA_DIR / "start_pictures.zip"
```

```python
model = keras.applications.ResNet50(
    weights="imagenet",
    include_top=True
)

print("Modell erfolgreich geladen!")
print("Anzahl Parameter:", model.count_params())
```

```python
image = extract_picture_file("brille")

image_array = vectorize_image(image)
```

```python
decoded_predictions = predict_picture(model, image_array)

for imagenet_id, label, probability in decoded_predictions[0]:
    print(f"{label}: {probability:.2%}")
```
