```python
from pathlib import Path
import sys
import numpy as np

BASE_DIR = Path().cwd().parent.parent
PICTURE_DATA_DIR = BASE_DIR / "picture_data"
DATASET_DIR = PICTURE_DATA_DIR / "start_bilder_dataset"
TRAIN_DIR = DATASET_DIR / "train"
VALIDATION_DIR = DATASET_DIR / "validation"
TEST_DIR = DATASET_DIR / "test"

EXPERIMENTS_DIR = BASE_DIR / "experimental_models"
CASE_DIR = EXPERIMENTS_DIR / "just_start_bilder"
FEATURE_DIR = CASE_DIR / "features"

CLASS_NAMES = [
    "brille",
    "flasche",
    "geldboerse",
    "handy",
    "schuh",
    "tasse"
]

VERSION = 1

sys.path.append(str(BASE_DIR))

from data_handling.manage_data import create_dataset, extract_features
from model.build_model import build_feature_extractor
```

```python
feature_extractor = build_feature_extractor()
```

```python
train_ds = create_dataset(TRAIN_DIR, CLASS_NAMES, batch_size=16)
validation_ds = create_dataset(VALIDATION_DIR, CLASS_NAMES, batch_size=16)
test_ds = create_dataset(TEST_DIR, CLASS_NAMES, batch_size=16)
```

```python
feature_extractor = build_feature_extractor()

X_train, y_train = extract_features(feature_extractor, train_ds)
X_validation, y_validation = extract_features(feature_extractor, validation_ds)
X_test, y_test = extract_features(feature_extractor, test_ds)


np.savez_compressed(
    FEATURE_DIR / f"features_{VERSION}.npz",
    X_train=X_train,
    y_train=y_train,
    X_validation=X_validation,
    y_validation=y_validation,
    X_test=X_test,
    y_test=y_test
)
```
