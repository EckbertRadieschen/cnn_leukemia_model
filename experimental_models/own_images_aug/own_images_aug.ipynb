```python
from pathlib import Path
import sys

BASE_DIR = Path().cwd().parent.parent
PICTURE_DATA_DIR = BASE_DIR / "picture_data"
DATASET_DIR = PICTURE_DATA_DIR / "dataset"
TRAIN_DIR = DATASET_DIR / "train"
VALIDATION_DIR = DATASET_DIR / "validation"
TEST_DIR = DATASET_DIR / "test"

EXPERIMENTS_DIR = BASE_DIR / "experimental_models"
CASE_DIR = EXPERIMENTS_DIR / "own_images_aug"
FEATURE_DIR = CASE_DIR / "features"
MODELS_DIR = CASE_DIR / "models"

VERSION = 1

CLASS_NAMES = [
    "brille",
    "flasche",
    "geldboerse",
    "handy",
    "hund",
    "kirche",
    "schuh",
    "tasse"
]

sys.path.append(str(BASE_DIR))

from data_handling.manage_data import load_features
from model.build_model import build_model, save_model
from evaluation.evaluation import evaluate_model

from model.training_object_classes import EpochInfo, get_early_stopping
```

```python
X_train, y_train, X_validation, y_validation, X_test, y_test = load_features(FEATURE_DIR, VERSION)
```

```python
model = build_model(output_count=8, learning_rate=0.00005)
```

```python
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_validation, y_validation),
    epochs=30,
    shuffle=True,
    callbacks=[EpochInfo(), get_early_stopping()],
    verbose=1
)
```

```python
evaluate_model(model, X_validation, y_validation, CLASS_NAMES)
```

```python
save_model(MODELS_DIR, model, "just_aug_very_slow")
```
